```python
# Dataset 1: Make + Model ko combine karke ek match-key banao
df1['match_key'] = (df1['Make'] + ' ' + df1['Model']).str.lower().str.strip()

# Dataset 2: Car_Name already "Brand Model" jaisa hai, seedha use karo
df2['match_key'] = df2['Car_Name'].str.lower().str.strip()

# Kitne unique match keys dono me hain
print("Dataset 1 unique match_keys:", df1['match_key'].nunique())
print("Dataset 2 unique match_keys:", df2['match_key'].nunique())

# Kitne overlap kar rahe hain (direct exact match)
common = set(df1['match_key']) & set(df2['match_key'])
print("Directly matching models:", len(common))
```

```python
# Jo Dataset 2 ke match_keys direct match nahi hue unmein se kuch dekho
unmatched = df2[~df2['match_key'].isin(df1['match_key'])]
print(unmatched[['Brand', 'Car_Name', 'match_key']].head(15))
```

```python
# Maruti wali rows Dataset 1 me kaise dikhti hain
print(df1[df1['Make'].str.contains('Maruti', case=False, na=False)][['Make', 'Model', 'match_key']].head(5))
```

```python
# Sirf Model (bina Make ke) use karke naya match_key banao
df1['match_key_v2'] = df1['Model'].str.lower().str.strip()

# Dataset 2: Car_Name se pehla word (brand) hata do, baaki model treat karo
df2['match_key_v2'] = df2['Car_Name'].str.replace(r'^(Maruti Suzuki|Maruti|Tata|Hyundai|Mahindra|Kia|Toyota|Honda|MG|Skoda|Volkswagen|Renault|Nissan|Citroen|Jeep)\s+', '', regex=True, case=False).str.lower().str.strip()

common_v2 = set(df1['match_key_v2']) & set(df2['match_key_v2'])
print("New matching count:", len(common_v2))
```

```python
from rapidfuzz import process, fuzz

# Dataset 1 ka full "Make Model" string (unique list)
df1['full_name'] = (df1['Make'] + ' ' + df1['Model']).str.lower().str.strip()
df1_names = df1['full_name'].unique().tolist()

# Har Dataset 2 car ke liye best match dhoondo Dataset 1 me
def find_best_match(name, choices, threshold=85):
    result = process.extractOne(name, choices, scorer=fuzz.token_sort_ratio)
    if result and result[1] >= threshold:
        return result[0]  # matched name
    return None

df2['matched_full_name'] = df2['match_key'].apply(lambda x: find_best_match(x, df1_names))

matched_count = df2['matched_full_name'].notna().sum()
print(f"Fuzzy matched: {matched_count} out of {len(df2)}")
```

```python
df2['matched_full_name_v2'] = df2['match_key'].apply(
    lambda x: find_best_match(x, df1_names, threshold=88)
)
```

```python
def find_best_match_v2(name, choices, threshold=88):
    result = process.extractOne(name, choices, scorer=fuzz.token_set_ratio)
    if result and result[1] >= threshold:
        return result[0]
    return None

df2['matched_full_name_v2'] = df2['match_key'].apply(lambda x: find_best_match_v2(x, df1_names))

matched_count_v2 = df2['matched_full_name_v2'].notna().sum()
print(f"Fuzzy matched (v2): {matched_count_v2} out of {len(df2)}")
```

```python
# Sirf wo rows dikhao jo match hue, side-by-side compare karne ke liye
matched_preview = df2[df2['matched_full_name_v2'].notna()][['Car_Name', 'matched_full_name_v2']]
print(matched_preview.to_string())
```

```python
import re

def has_conflict(name1, name2):
    # Rule 1: agar dono me numbers hain, wo match hone chahiye
    nums1 = set(re.findall(r'\d+', name1))
    nums2 = set(re.findall(r'\d+', name2))
    if nums1 and nums2 and nums1 != nums2:
        return True
    
    # Rule 2: EV/Electric keyword sirf ek taraf hona chahiye ni
    ev_keywords = ['electric', ' ev', 'e-tron', 'eq', 'ix', ' se']
    name1_has_ev = any(k in f' {name1} ' for k in ev_keywords)
    name2_has_ev = any(k in f' {name2} ' for k in ev_keywords)
    if name1_has_ev != name2_has_ev:
        return True
    
    return False

# Conflict check lagao existing matches pe
df2['has_conflict'] = df2.apply(
    lambda row: has_conflict(str(row['Car_Name']).lower(), str(row['matched_full_name_v2'])) 
    if pd.notna(row['matched_full_name_v2']) else False, 
    axis=1
)

print("Conflicts found:", df2['has_conflict'].sum())
print(df2[df2['has_conflict']][['Car_Name', 'matched_full_name_v2']].to_string())
```

```python
def has_conflict_v2(name1, name2):
    nums1 = set(re.findall(r'\d+', name1))
    nums2 = set(re.findall(r'\d+', name2))
    if nums1 and nums2 and nums1 != nums2:
        return True
    
    # \b use karo taaki "series" ke andar "se" match na ho
    ev_pattern = r'\b(electric|ev|e-tron|eq[a-z]?|ix\d|urus se)\b'
    name1_has_ev = bool(re.search(ev_pattern, name1))
    name2_has_ev = bool(re.search(ev_pattern, name2))
    if name1_has_ev != name2_has_ev:
        return True
    
    return False

df2['has_conflict_v2'] = df2.apply(
    lambda row: has_conflict_v2(str(row['Car_Name']).lower(), str(row['matched_full_name_v2'])) 
    if pd.notna(row['matched_full_name_v2']) else False, 
    axis=1
)

print("Conflicts (v2):", df2['has_conflict_v2'].sum())
print(df2[df2['has_conflict_v2']][['Car_Name', 'matched_full_name_v2']].to_string())
```

```python
# Guard ne jo pakde (genuine EV/spec conflicts)
auto_excluded = df2[df2['has_conflict_v2']]['Car_Name'].tolist()

# Guard se bach gaye, manually dekhe hue galat matches
manual_exclude = ['Range Rover', 'Porsche Taycan', 'Aston Martin DBX']

all_excluded = set(auto_excluded) | set(manual_exclude)
print("Total excluded:", len(all_excluded))

# Final clean matched dataset — sirf wahi rows rakho jo match hue AUR exclude list me nahi
df2_clean = df2[
    df2['matched_full_name_v2'].notna() & 
    ~df2['Car_Name'].isin(all_excluded)
].copy()

print("Final clean matched rows:", len(df2_clean))
```

```python
# df2_clean me se sirf zaroori columns lo merge ke liye
df2_to_merge = df2_clean[['matched_full_name_v2', 'Rating', 'Safety_Stars', 'Mileage_kmpl', 'Sales_FY2024']].copy()
df2_to_merge = df2_to_merge.rename(columns={'matched_full_name_v2': 'full_name'})

# df1 (specs) ke saath left-merge karo — df1 ki saari rows preserve rahengi
df_merged = df1.merge(df2_to_merge, on='full_name', how='left')

print("Merged shape:", df_merged.shape)
print()
print("Rows with mileage data:", df_merged['Mileage_kmpl'].notna().sum())
print("Rows with safety data:", df_merged['Safety_Stars'].notna().sum())
print()
df_merged.head()
```

```python
# Check karo kya full_name duplicate hai matched dataset me
dup_check = df2_to_merge['full_name'].value_counts()
duplicates = dup_check[dup_check > 1]
print("Duplicate full_names:", len(duplicates))
print(duplicates)
```

```python
# Dedup: duplicate full_name me se pehli row rakho
df2_to_merge_dedup = df2_to_merge.drop_duplicates(subset='full_name', keep='first')
print("Before dedup:", len(df2_to_merge), "After dedup:", len(df2_to_merge_dedup))

# Merge dobara karo, dedup wale data se
df_merged = df1.merge(df2_to_merge_dedup, on='full_name', how='left')

print()
print("Merged shape:", df_merged.shape)
print("Rows with mileage data:", df_merged['Mileage_kmpl'].notna().sum())
print("Rows with safety data:", df_merged['Safety_Stars'].notna().sum())
```

```python
# Price: ₹ symbol, commas, extra decimals hatao, number banao
df_merged['Price'] = (
    df_merged['Ex-Showroom_Price']
    .astype(str)
    .str.replace('₹', '', regex=False)
    .str.replace(',', '', regex=False)
    .str.strip()
    .astype(float)
)

# Columns ko clean, meaningful naam do
df_merged = df_merged.rename(columns={
    'Make': 'brand',
    'Model': 'model',
    'Variant': 'variant',
    'Price': 'price',
    'Displacement': 'engine_cc',
    'Fuel_Type': 'fuel_type',
    'Body_Type': 'body_type',
    'Seating_Capacity': 'seating_capacity',
    'Type': 'transmission',
    'Power.1': 'power_bhp',
    'Torque.1': 'torque_nm',
    'Mileage_kmpl': 'mileage',
    'Safety_Stars': 'safety_rating',
    'Ground_Clearance': 'ground_clearance'  # agar exist kare
})

# Check karo price sahi ban gaya
print(df_merged[['brand', 'model', 'variant', 'price', 'fuel_type', 'transmission', 'mileage', 'safety_rating']].head(10))
print()
print("Price stats:")
print(df_merged['price'].describe())
```

```python
# Pehle dekho kitna missing hai body_type ke hisaab se
print("Missing mileage %:")
print(df_merged.groupby('body_type')['mileage'].apply(lambda x: x.isna().mean() * 100).round(1))
print()
print("Missing safety_rating %:")
print(df_merged.groupby('body_type')['safety_rating'].apply(lambda x: x.isna().mean() * 100).round(1))
```

```python
# body_type ko simplify: comma ke baad wala hata do (sirf pehla/primary type rakho)
df_merged['body_type_clean'] = df_merged['body_type'].astype(str).str.split(',').str[0].str.strip()

print(df_merged['body_type_clean'].value_counts())
```

```python
# Step A: per body_type_clean median se fill karne ki koshish
df_merged['mileage'] = df_merged.groupby('body_type_clean')['mileage'].transform(
    lambda x: x.fillna(x.median())
)
df_merged['safety_rating'] = df_merged.groupby('body_type_clean')['safety_rating'].transform(
    lambda x: x.fillna(x.median())
)

# Step B: jo abhi bhi NaN bache (poora group hi missing tha), unhe overall median se fill karo
df_merged['mileage'] = df_merged['mileage'].fillna(df_merged['mileage'].median())
df_merged['safety_rating'] = df_merged['safety_rating'].fillna(df_merged['safety_rating'].median())

print("Remaining missing mileage:", df_merged['mileage'].isna().sum())
print("Remaining missing safety:", df_merged['safety_rating'].isna().sum())
```

```python
# Track karo kaunsi rows imputed thi (asli merge se pehle NaN thi)
# Note: humne already overwrite kar diya hai, isliye ye flag pehle hi bana lena chahiye tha —
# lekin humare paas abhi bhi 'df2_to_merge_dedup' se pata chal sakta hai kaunsi rows match hui thi

df_merged['mileage_is_imputed'] = ~df_merged['full_name'].isin(df2_to_merge_dedup['full_name'])
df_merged['safety_is_imputed'] = df_merged['mileage_is_imputed']  # same source, same flag

print("Imputed mileage rows:", df_merged['mileage_is_imputed'].sum())
print("Real mileage rows:", (~df_merged['mileage_is_imputed']).sum())
```

```python
final_columns = [
    'brand', 'model', 'variant', 'price', 'fuel_type', 'transmission',
    'engine_cc', 'power_bhp', 'torque_nm', 'seating_capacity',
    'body_type_clean', 'mileage', 'safety_rating',
    'mileage_is_imputed', 'safety_is_imputed'
]

df_final = df_merged[final_columns].rename(columns={'body_type_clean': 'body_type'})

# Save karo processed folder me
df_final.to_csv('../data/processed/cars_processed.csv', index=False)

print("Saved! Final shape:", df_final.shape)
df_final.head()
```
